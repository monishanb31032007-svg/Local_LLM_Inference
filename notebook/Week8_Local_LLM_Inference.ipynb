## 2. Introduction to Local LLM Inference

Large Language Models (LLMs) can generate human-like text,
answer questions, explain concepts, and write programs.

Ollama allows us to download and run supported language models
on our local computer.

Local LLM inference can improve privacy, reduce dependence on
cloud APIs, and allow experimentation without paying for each
API request. However, local models still require sufficient
memory and computing resources.

In this project, we use Ollama and Python to generate responses
for prompts from different domains and compare model outputs.

## 3. Ollama Installation and Verification

Ollama was installed on the local Windows computer.

The following terminal commands are used to verify the
installation and check the available models:

- `ollama --version`
- `ollama list`

```python
import subprocess

commands = [
    ["ollama", "--version"],
    ["ollama", "list"]
]

for command in commands:
    print("\nCommand:", " ".join(command))

    try:
        result = subprocess.run(
            command,
            capture_output=True,
            text=True,
            timeout=30
        )

        print(result.stdout)

        if result.stderr:
            print(result.stderr)

    except Exception as error:
        print("Error:", error)
```

## 4. Available Local Models

The local models used in this project are:

1. Llama 3.2
2. Gemma 2B

Llama 3.2 is used to generate responses for the initial
prompt dataset. Gemma 2B is used for comparison.

Both models are accessed through Ollama. Their response
quality, length, relevance, and accuracy are evaluated.

Note: The models must be downloaded before they can be used.

## 5. Python Library Import

The following Python libraries are used:

- pandas: To load and analyse CSV datasets.
- ollama: To communicate with the local language model.
- pathlib: To manage project file paths.
- IPython.display: To display results in the notebook.

```python
import pandas as pd
from pathlib import Path
from ollama import chat
from IPython.display import display

print("Libraries imported successfully.")
```

## 6. Loading the Prompt Dataset

The prompts.csv file contains ten prompts from different
domains, including education, healthcare, coding, travel,
business, environment, technology, science, problem solving,
and creative writing.

The dataset is loaded using pandas for further processing.

```python
BASE_DIR = Path.cwd().parent

DATASET_DIR = BASE_DIR / "dataset"

PROMPTS_FILE = DATASET_DIR / "prompts.csv"
RESPONSES_FILE = DATASET_DIR / "llm_responses.csv"

prompts_df = pd.read_csv(
    PROMPTS_FILE,
    encoding="utf-8-sig"
)

print("Total prompts:", len(prompts_df))

display(prompts_df)
```

## 7. Generating LLM Responses

The Ollama Python library connects to the local Llama 3.2 model.

Each prompt is sent to the model, and the generated response
is collected for analysis.

The responses are saved in llm_responses.csv for future use.

```python
MODEL_NAME = "llama3.2"

results = []

for _, row in prompts_df.iterrows():

    prompt = row["prompt"]

    try:
        response = chat(
            model=MODEL_NAME,
            messages=[
                {
                    "role": "user",
                    "content": prompt
                }
            ]
        )

        answer = response.message.content

        results.append({
            "prompt_id": row["prompt_id"],
            "domain": row["domain"],
            "model": MODEL_NAME,
            "prompt": prompt,
            "response": answer,
            "response_word_count": len(answer.split()),
            "status": "Success"
        })

        print(f"Completed: {row['prompt_id']}")

    except Exception as error:
        results.append({
            "prompt_id": row["prompt_id"],
            "domain": row["domain"],
            "model": MODEL_NAME,
            "prompt": prompt,
            "response": "",
            "response_word_count": 0,
            "status": f"Error: {error}"
        })

        print(f"Failed: {row['prompt_id']} - {error}")

results_df = pd.DataFrame(results)

results_df.to_csv(
    RESPONSES_FILE,
    index=False,
    encoding="utf-8"
)

print("\nResponse generation completed.")

display(results_df)
```

## 8. Displaying and Analysing Results

The generated responses are analysed using pandas.

The analysis includes:
- The total number of prompts.
- The number of successful responses.
- The average response length.
- The number of responses for each domain.

```python
# Display the generated responses
display(
    results_df[
        [
            "prompt_id",
            "domain",
            "response_word_count",
            "status"
        ]
    ]
)

# Count successful responses
successful = (
    results_df["status"] == "Success"
).sum()

print("Total prompts:", len(results_df))
print("Successful responses:", successful)
print("Failed responses:", len(results_df) - successful)

# Calculate average response length for successful responses
successful_df = results_df[
    results_df["status"] == "Success"
]

if not successful_df.empty:
    average_words = successful_df[
        "response_word_count"
    ].mean()

    print(
        "Average response length:",
        round(average_words, 2),
        "words"
    )

    print("\nResponses by domain:")

    display(
        successful_df.groupby("domain").size()
        .reset_index(name="number_of_responses")
    )
```

## 9. Comparing Llama and Gemma

Llama 3.2 and Gemma 2B are evaluated using the same ten prompts.

The comparison considers:

1. Response quality and clarity.
2. Response length.
3. Accuracy.
4. Relevance to the prompt.

Both models receive identical prompts to make the comparison
fairer. Their generated responses are saved in
response_comparison.csv.

Accuracy is assessed by checking the answers against
appropriate reference information. Response length alone
does not determine response quality.

```python
COMPARISON_FILE = DATASET_DIR / "response_comparison.csv"

comparison_results = []

MODEL_NAMES = ["llama3.2", "gemma2:2b"]

for _, row in prompts_df.iterrows():

    prompt_id = row["prompt_id"]
    domain = row["domain"]
    prompt = row["prompt"]

    for model_name in MODEL_NAMES:

        try:
            response = chat(
                model=model_name,
                messages=[
                    {
                        "role": "user",
                        "content": prompt
                    }
                ]
            )

            answer = response.message.content

            comparison_results.append({
                "prompt_id": prompt_id,
                "domain": domain,
                "model": model_name,
                "prompt": prompt,
                "response": answer,
                "response_word_count": len(answer.split()),
                "status": "Success"
            })

            print(f"{model_name}: {prompt_id} completed")

        except Exception as error:

            comparison_results.append({
                "prompt_id": prompt_id,
                "domain": domain,
                "model": model_name,
                "prompt": prompt,
                "response": "",
                "response_word_count": 0,
                "status": f"Error: {error}"
            })

            print(f"{model_name}: {prompt_id} failed - {error}")


comparison_df = pd.DataFrame(comparison_results)

comparison_df.to_csv(
    COMPARISON_FILE,
    index=False,
    encoding="utf-8"
)

print("\nComparison dataset saved.")

display(
    comparison_df[
        [
            "prompt_id",
            "domain",
            "model",
            "response_word_count",
            "status"
        ]
    ]
)
```

```python
successful_comparisons = comparison_df[
    comparison_df["status"] == "Success"
]

summary = (
    successful_comparisons
    .groupby("model")
    .agg(
        successful_responses=("prompt_id", "count"),
        average_response_words=("response_word_count", "mean")
    )
    .reset_index()
)

summary["average_response_words"] = (
    summary["average_response_words"].round(2)
)

display(summary)
```
